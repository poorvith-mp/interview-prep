# AI Interview Prep Coach — Development Notebook
Building an AI-powered interview coach using Ollama + Mistral

```python
import ollama
```

```python
# Step 1: Basic test — ask mistral a simple interview question
response = ollama.chat(
    model='mistral',
    messages=[
        {
            'role': 'user',
            'content': 'What is the difference between a list and a tuple in Python?'
        }
    ]
)

print(response['message']['content'])
```

```python
# Step 2: Accept dynamic user input
user_input = input("Ask an interview question: ")

response = ollama.chat(
    model='mistral',
    messages=[
        {
            'role': 'user',
            'content': user_input
        }
    ]
)

print(response['message']['content'])
```

```python
# Step 3: Build a reusable helper function
def ask_ai(question):
    response = ollama.chat(
        model='mistral',
        messages=[
            {
                'role': 'user',
                'content': question
            }
        ]
    )
    return response['message']['content']
```

```python
# Step 4: Add a system prompt to make it role-aware
def interview_coach(job_role, question):
    response = ollama.chat(
        model='mistral',
        messages=[
            {
                'role': 'system',
                'content': f"""You are an expert interview coach specializing in {job_role} roles.
                rules:
                    - Be encouraging and constructive
                    - Give clear, structured answers with bullet points
                    - For technical questions, provide code examples when relevant
                    - For behavioral questions, use the STAR method
                    - Keep answers concise and interview-ready
                    - End with a quick tip to improve the answer
                """
            },
            {
                'role': 'user',
                'content': question
            }
        ]
    )
    return response['message']['content']
```

```python
# Step 5: Test the interview coach
interview_coach("Data Scientist", "Tell me about yourself")
```
